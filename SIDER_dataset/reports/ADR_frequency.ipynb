```python
ADR_frequency = pd.read_csv("./ADR_frequency.csv", index_col=False)
ADR_frequency_above_10 = ADR_frequency[ADR_frequency["Count"] > 10]
ADR_frequency_above_10.to_csv("./ADR_frequency_above_10.csv", index=False)
print(len(ADR_frequency_above_10))
chunk_size = math.floor(len(ADR_frequency_above_10) / 3)
print(chunk_size)
high_freq = ADR_frequency_above_10.iloc[0:chunk_size]
# print(f"high_freq rows: {len(high_freq)}")
print(f"high_freq range: {high_freq['Frequency'].min()} - {high_freq['Frequency'].max()}")
print(f"high_freq mean: {high_freq['Frequency'].mean()}")
mid_freq = ADR_frequency_above_10.iloc[chunk_size:2 * chunk_size]
# print(f"mid_freq rows: {len(mid_freq)}")
print(f"mid_freq range: {mid_freq['Frequency'].min()} - {mid_freq['Frequency'].max()}")
print(f"mid_freq mean: {mid_freq['Frequency'].mean()}")
low_freq = ADR_frequency_above_10.iloc[2 * chunk_size:3 * chunk_size]
# print(f"low_freq rows: {len(low_freq)}")
print(f"low_freq range: {low_freq['Frequency'].min()} - {low_freq['Frequency'].max()}")
print(f"low_freq mean: {low_freq['Frequency'].mean()}")
```
