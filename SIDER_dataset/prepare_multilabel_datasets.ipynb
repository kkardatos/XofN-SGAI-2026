```python
import sys, os

from SIDER_dataset.libraries.PCT_library import save_to_arff
from SIDER_dataset.libraries.prepare_dataset_library import get_ten_labels_mid

sys.path.insert(0, os.path.abspath("C:/Users/Voror/Projects/Personal/sep"))
%load_ext autoreload
%autoreload 2
from SIDER_dataset.libraries.prepare_dataset_library import *

high_freq_labels_with_names, high_freq_labels = get_high_freq_labels()
mid_freq_labels_with_names, mid_freq_labels = get_mid_freq_labels()
low_freq_labels_with_names, low_freq_labels = get_low_freq_labels()
cardiac_labels_with_names, cardiac_labels = get_cardiac_labels()
ten_labels_with_names, ten_labels = get_ten_labels()
ten_labels_mid_with_names, ten_labels_mid = get_ten_labels_mid()
labels = [high_freq_labels, mid_freq_labels, low_freq_labels, cardiac_labels, ten_labels, ten_labels_mid]

for label_set in labels:
    print(f"{label_set}: {get_labels_name(label_set)}")

dataset_names = ["fingerprint", "CPI", "CPI+fingerprint"]
feature_binary_threshold = 500
feature_drop_threshold = 10
```

```python
for label_set in labels:
    for dataset_name in dataset_names:
        print(
            f"Processing {dataset_name} for {len(label_set)} labels: {label_set} with name: {get_labels_name(label_set)}")
        dataset = get_dataset(dataset_name, label_set)
        parent_path = "datasets/clean_multi_label_datasets"
        dataset_path = f"{parent_path}/{dataset_name}_{get_labels_name(label_set)}"

        current_df = dataset.drop(columns=["CID"])
        if dataset_name == "CPI" or dataset_name == "CPI+fingerprint":
            current_df = binarize_features(current_df,
                                           get_cpis(current_df),
                                           feature_binary_threshold)
        original_feature_frequency = get_feature_1_counts(dataset, label_set)
        binary_feature_columns = get_features(current_df, label_set)
        (
            current_df,
            current_labels,
            current_features,
            current_binary_features,
            current_prob_features,
        ) = drop_uninformative_features(
            current_df,
            feature_drop_threshold,
            label_set,
            binary_feature_columns,
        )

        current_df = remove_inconsistent_duplicates(
            current_df, current_features, current_labels
        )
        for label in label_set:
            current_df[label] = current_df.pop(label)

        csv_path = f"{dataset_path}.csv"
        current_df.to_csv(csv_path, index=False)
        print(csv_path + "\n")
```
