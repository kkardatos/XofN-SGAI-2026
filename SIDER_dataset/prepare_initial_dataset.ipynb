```python
%load_ext autoreload
%autoreload 2

# Prepare full dataset
import pandas as pd
import pubchempy as pcp
import numpy as np


def get_fingerprint(cid: str):
    code = cid[4:]
    pubchem_compound_data = pcp.Compound.from_cid(int(code))
    fingerprint = bin(int(pubchem_compound_data.fingerprint, 16))
    return fingerprint


def add_row(df, cid, fingerprint):
    new_row = [cid] + list(fingerprint)
    df.loc[len(df)] = new_row
    return df


from SIDER_dataset.libraries.prepare_dataset_library import *
```
