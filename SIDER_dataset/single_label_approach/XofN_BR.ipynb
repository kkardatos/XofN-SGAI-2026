```python
# Implement automated x-of-n on fingerprints dataset
# first implement for single label
# then apply multi-label
%load_ext autoreload
%autoreload 2

from library import *
from SIDER_dataset.libraries.XofN_library import *
import itertools
import json
from SIDER_dataset.libraries.PCT_library import save_to_arff
```

```python
# Set ADR to predict and scoring
# se_C0027497 1179 C0027497 ['Nausea'] 0.845
# se_C0018681 1092 C0018681 ['Headache'] 0.783
# se_C0011603 1070 C0011603 ['Dermatitis'] 0.767
# se_C0015230 1062 C0015230 ['Rash'] 0.761
# se_C0042963 1060 C0042963 ['Vomiting'] 0.760
# se_C0012833 1014 C0012833 ['Dizziness'] 0.727

# Additional ADRs
# se_C0027769 431 ['Nervousness'] 0.309
# se_C0020580 354 ['Hypoaesthesia'] 0.254
# se_C0014457 286 ['Eosinophilia'] 0.205
# se_C0017181 214 ['Gastrointestinal haemorrhage'] 0.154
# se_C0151763 146 ['Hepatocellular injury'] 0.105
# se_C0038358 70 ['Gastric ulcer'] 0.050

labels_to_names = {
    # original ADRs
    "se_C0027497": "Nausea",
    "se_C0018681": "Headache",
    "se_C0011603": "Dermatitis",
    "se_C0015230": "Rash",
    "se_C0042963": "Vomiting",
    "se_C0012833": "Dizziness",
    # additional ADRs
    "se_C0027769": "Nervousness",
    "se_C0020580": "Hypoaesthesia",
    "se_C0014457": "Eosinophilia",
    "se_C0017181": "Gastrointestinal haemorrhage",
    "se_C0151763": "Hepatocellular injury",
    "se_C0038358": "Gastric ulcer",
}

labels = [ADR for ADR in labels_to_names]
dataset_name = "fingerprint"
dataset = get_dataset(dataset_name, labels)

k = 10
model_type = "RF"
random_state = 42
binary_feature_columns = get_fs(dataset)
feature_threshold = 10
dataset
```

```python
# Create single label datasets
all_labels_cv_results = {}
for current_label in labels:
    print(f"\n--- Creating dataset for label: {current_label} ---")
    current_df = dataset.drop(
        columns=[label for label in labels if label != current_label]
    )
    current_df = current_df.drop(columns=["CID"])

    (
        current_df,
        current_labels,
        current_features,
        current_binary_features,
        current_prob_features,
    ) = drop_uninformative_features(
        current_df,
        feature_threshold,
        [current_label],
        binary_feature_columns,
    )

    current_df = remove_inconsistent_duplicates(
        current_df, current_features, current_labels
    )
    current_df[current_label] = current_df.pop(current_label)
    csv_path = f"clean_single_label_datasets/{current_label}.csv"
    current_df.to_csv(csv_path, index=False)
    save_to_arff(csv_path, [current_label])
```

```python
# Ungrouped
all_labels_cv_results = {}
for current_label in labels:
    print(f"\n--- Processing label: {current_label} ---")
    current_df = pd.read_csv(f"clean_single_label_datasets/{current_label}.csv")
    kf = KFold(n_splits=k, shuffle=True, random_state=random_state)

    for fold, (train_idx, test_idx) in enumerate(kf.split(current_df)):
        print(f"\nFold {fold + 1}/{k}")

        train_dataset = current_df.iloc[train_idx]
        test_dataset = current_df.iloc[test_idx]

        model = train(model_type, train_dataset, [current_label])
        performance_df = test(model, test_dataset, [current_label])
        performance_df["fold"] = fold + 1
        if current_label not in all_labels_cv_results:
            all_labels_cv_results[current_label] = []
        all_labels_cv_results[current_label].append(performance_df)

    all_labels_cv_results[current_label] = pd.concat(
        all_labels_cv_results[current_label]
    ).reset_index()
all_labels_cv_results
```

```python
params = {
    "label": labels,
    "training_algorithm": ["DT"],
    "eval_criterion": ["roc_auc"],
    "max_size": [5],
}
# ranking_criteria = ["roc_auc", "MDI", "MDA", "AUROC_MDI_MDA", "random"]
ranking_criteria = ["MDI"]
include_original_features_options = [True, False]
all_labels_cv_results = []
mean_labels_results = []
for combo in itertools.product(*params.values()):
    # Setting parameters for run
    config = dict(zip(params.keys(), combo))
    current_label = config["label"]
    training_algorithm = config["training_algorithm"]
    eval_criterion = config["eval_criterion"]
    max_size = config["max_size"]
    run_config = f"\n--- Running with label:'{current_label}' training_algorithm:'{training_algorithm}' eval_criterion:'{eval_criterion}' max_size:'{max_size}' ---"
    print(run_config)
    run_config_name = "_".join(
        [
            current_label,
            training_algorithm,
            eval_criterion,
            str(max_size),
        ]
    )
    logging_path = f"datasets/XofN_datasets/auto_groupings/{run_config_name}_logs.txt"
    print(f"Logs can be found in {logging_path}.")
    logger = get_logger(logging_path)
    logger.info(run_config_name)

    # Load dataset
    current_df = pd.read_csv(f"clean_single_label_datasets/{current_label}.csv")

    kf = KFold(n_splits=k, shuffle=True, random_state=random_state)
    performance_dfs = []

    for fold, (train_idx, test_idx) in enumerate(kf.split(current_df)):
        print(f"\nFold {fold + 1}/{k}")
        train_dataset = current_df.iloc[train_idx]
        test_dataset = current_df.iloc[test_idx]

        learn_set, val_set = train_test_split(
            train_dataset, test_size=0.2, random_state=random_state
        )
        feature_perfs = calculate_feature_perfs(learn_set, val_set, current_label)
        for ranking_criterion in ranking_criteria:
            rankings = generate_feature_rankings(feature_perfs, ranking_criterion)
            print(rankings)

            XofN_groupings = generate_XofN_list(
                training_algorithm,
                learn_set,
                val_set,
                rankings,
                eval_criterion,
                max_size,
                current_label,
                logger,
            )

            for include_original_features in include_original_features_options:
                current_train_dataset = group_features(
                    train_dataset,
                    [current_label],
                    XofN_groupings,
                    include_original_features,
                )

                current_test_dataset = group_features(
                    test_dataset,
                    [current_label],
                    XofN_groupings,
                    include_original_features,
                )

                if include_original_features:
                    importances = get_feature_importance(current_train_dataset, current_label, "roc_auc", k)
                    features_to_keep = get_top_k_features(current_train_dataset, importances, 0.6)
                    current_train_dataset = current_train_dataset[features_to_keep + [current_label]]
                    current_test_dataset = current_test_dataset[features_to_keep + [current_label]]

                model = train(model_type, current_train_dataset, [current_label])
                performance_df = test(model, current_test_dataset, [current_label])
                performance_df["fold"] = fold + 1
                performance_df["ranking_criterion"] = ranking_criterion
                performance_df["training_algorithm"] = training_algorithm
                performance_df["eval_criterion"] = eval_criterion
                performance_df["include_original_features"] = (
                    "with_org" if include_original_features else "no_org"
                )
                performance_df["max_size"] = str(max_size)
                performance_dfs.append(performance_df)
                print(performance_df.to_string())

    all_cv_results = pd.concat(performance_dfs).reset_index()
    all_labels_cv_results.append(all_cv_results)

for handler in logging.root.handlers[:]:
    handler.close()
    logging.root.removeHandler(handler)

all_labels_cv_results_df = pd.concat(all_labels_cv_results).reset_index()
all_labels_cv_results_df
```

```python
# how many XofN features are selected from all features
import re
import pandas as pd

# Read file
with open("../datasets/XofN_datasets/logs", "r", encoding="utf-8", errors="ignore") as f:
    text = f.read()

# Regex pattern to capture both numbers
pattern = r"Selecting .*?features:\s*(\d+).*?\n(\d+)\s+XofN features selected"

# Find all matches
matches = re.findall(pattern, text)

# Convert to integers
data = [(int(total), int(selected)) for total, selected in matches]

# Put into a DataFrame
df = pd.DataFrame(data, columns=["TotalFeatures", "XofNSelected"])
df["percentage"] = df["XofNSelected"] / df["TotalFeatures"]
print(df)
print(df["percentage"].mean())
```

```python
def compare_with_baseline_results(xofn_results):
    xofn_results["no_org"] = xofn_results["no_org"].round(3)
    xofn_results["with_org"] = xofn_results["with_org"].round(3)
    br_bl_results = pd.read_csv("../reports/br_baseline_results.csv")
    br_bl_results["cv_auroc"] = br_bl_results["cv_auroc"].round(3)
    compare_results = pd.merge(br_bl_results, xofn_results, left_on=["ADR"], right_on=["index"])
    compare_results = compare_results.drop(columns=["index"])
    compare_results['max_perf'] = compare_results[['cv_auroc', 'no_org', 'with_org']].idxmax(axis=1)
    compare_results['max_perf_XofN'] = compare_results[['no_org', 'with_org']].idxmax(axis=1)
    print(compare_results['max_perf'].value_counts())
    print(compare_results['max_perf_XofN'].value_counts())
    return compare_results
```
