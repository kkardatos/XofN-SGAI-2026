```python
import pandas as pd
```

```python
meddra_all_se = pd.read_csv("meddra_all_se.tsv", sep="\t")
meddra_all_se = meddra_all_se.drop_duplicates()
print(len(meddra_all_se))
print(meddra_all_se["MedDRA_ADR_id"].unique().size)
meddra_all_se.head()
```

```python
meddra_all_se_drop_LLT = meddra_all_se[meddra_all_se["MedDRA_ADR_level"] == "PT"]
meddra_all_se_drop_LLT = meddra_all_se_drop_LLT.drop_duplicates()
print(len(meddra_all_se_drop_LLT))
print(meddra_all_se_drop_LLT["MedDRA_ADR_id"].unique().size)
meddra_all_se_drop_LLT.head()
```

```python
meddra_all_se_filtered = meddra_all_se_drop_LLT.drop(
    columns=[
        "CID0",
        "UMLS_ADR_id",
        "MedDRA_ADR_level",
    ]
)
meddra_all_se_filtered = meddra_all_se_filtered.drop_duplicates()
print("rows:", len(meddra_all_se_filtered))
print("drugs:", meddra_all_se_filtered["CID1"].unique().size)
print("ADRs:", meddra_all_se_filtered["MedDRA_ADR_id"].unique().size)
meddra_all_se_filtered.head()
```

```python
ADR_frequency = meddra_all_se_filtered["MedDRA_ADR_id"].value_counts()
ADR_frequency_df = ADR_frequency.reset_index()
ADR_frequency_df.columns = ["MedDRA_ADR_id", "Count"]
ADR_names = meddra_all_se_filtered[
    ["MedDRA_ADR_id", "MedDRA_ADR_name"]
].drop_duplicates()
ADR_frequency_df = ADR_frequency_df.merge(ADR_names, on="MedDRA_ADR_id", how="left")
ADR_frequency_df["Percentage"] = ADR_frequency_df["Count"] / len(ADR_frequency_df)
ADR_frequency_df.to_csv("SIDER_dataset_ADR_frequency", index=False)
ADR_frequency_df.head(10)
```

```python
SIDER_dataset = meddra_all_se_filtered.drop(columns=["MedDRA_ADR_name"])
SIDER_dataset = pd.get_dummies(
    SIDER_dataset, columns=["MedDRA_ADR_id"], prefix="se_", dtype=int
)
print("rows:", len(SIDER_dataset))
print("drugs:", SIDER_dataset["CID1"].unique().size)
SIDER_dataset = SIDER_dataset.groupby("CID1", as_index=False).max()
print("rows:", len(SIDER_dataset))
print("drugs:", SIDER_dataset["CID1"].unique().size)
SIDER_dataset.to_csv("./SIDER_dataset.csv", index=None)
SIDER_dataset.head()
```
