```python
import pandas as pd
```

```python
drug_atc = pd.read_csv("./SIDER-4.1/drug_atc.tsv", sep='\t')
print("drug_atc", drug_atc["CID"].unique().size)

drug_names = pd.read_csv("./SIDER-4.1/drug_names.tsv", sep='\t')
print("drug_names", drug_names["CID"].unique().size)

meddra_all_indications = pd.read_csv("./SIDER-4.1/meddra_all_indications.tsv", sep='\t')
print("meddra_all_indications", meddra_all_indications["CID"].unique().size)

meddra_all_se = pd.read_csv("./SIDER-4.1/meddra_all_se.tsv", sep='\t')
print("meddra_all_se", meddra_all_se["CID1"].unique().size)

meddra_freq = pd.read_csv("./SIDER-4.1/meddra_freq.tsv", sep='\t')
print("meddra_freq", meddra_freq["CID1"].unique().size)
```

```python
import requests
import json

# List of CIDs (Replace with your own list)
cids = meddra_freq["CID1"].unique()[0:10]

# Build API URL
url = f"https://pubchem.ncbi.nlm.nih.gov/rest/pug/compound/cid/{','.join(map(str, cids))}/property/Fingerprint2D/JSON"

# Fetch data
response = requests.get(url)

if response.status_code == 200:
    data = response.json()
    for compound in data['PropertyTable']['Properties']:
        print(f"CID: {compound['CID']}, Fingerprint: {compound['Fingerprint2D']}")
else:
    print("Error:", response.status_code)
```
