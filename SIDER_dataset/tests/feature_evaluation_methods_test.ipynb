```python
from SIDER_dataset.libraries.feature_evaluation_methods import feature_variance_reduction_scores
import numpy as np
import pandas as pd

np.random.seed(0)
n, m = 20, 6
X = pd.DataFrame({
    "f1": np.random.randint(0, 6, size=n),
    "f2": np.random.randint(0, 6, size=n),
    "f3": np.random.randint(0, 6, size=n),
    "f4": np.random.randint(0, 6, size=n)
})
Y = np.zeros((n, m), dtype=int)
Y[:, 0] = (X["f1"] > 2).astype(int)
Y[:, 1] = (X["f2"] < 3).astype(int)

scores = feature_variance_reduction_scores(X, Y, min_samples_leaf=2)
print("\nFinal scores:")
print(scores)
```
